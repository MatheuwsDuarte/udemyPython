```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
gas = pd.read_csv('gas_prices.csv')
gas
```

```python
plt.plot(gas['Year'], gas['Australia'])
plt.show()
```

```python
plt.plot(gas['Year'], gas['Australia'])
plt.plot(gas['Year'], gas['Italy'])
plt.plot(gas['Year'], gas['Germany'])
plt.show()
```

```python
plt.title('Valor do combustivel em US$')

plt.plot(gas['Year'], gas['Australia'], 'r.-', label = 'Australia')
plt.plot(gas['Year'], gas['Italy'], 'g.--', label = 'Italia')
plt.plot(gas['Year'], gas['Germany'], 'y.', label = 'Alemanha')
plt.plot(gas['Year'], gas['Japan'], 'b--', label = 'Japao')

plt.xticks(gas['Year'][::3]) #legenda do eixo x vai do começo ao fim de 3 em 3
plt.xlabel('Anos')
plt.ylabel('US$')

plt.legend()
plt.show()
```

```python
plt.title('Valor do combustivel em US$')

listaPaises = ['Australia', 'Estados Unidos', 'Canada', 'Mexico']

for pais in gas:
    if pais in listaPaises:
        plt.plot(gas['Year'], gas[pais], '.-', label = pais)


plt.xticks(gas['Year'][::3]) #legenda do eixo x vai do começo ao fim de 3 em 3
plt.xlabel('Anos')
plt.ylabel('US$')

plt.legend()
plt.savefig('paises.png')
plt.savefig('paises.pdf')
plt.show()
```
